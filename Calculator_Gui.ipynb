## Import Layer

```python
import sys
from PySide6.QtWidgets import QApplication, QWidget, QLineEdit, QVBoxLayout, QPushButton, QHBoxLayout
from PySide6.QtCore import Qt
```

## Code Layer

```python
app = QApplication.instance()
if app is None:
    app = QApplication([])

window = QWidget()

window.setWindowTitle("Calculator Gui")

num1 = QLineEdit()
num2 = QLineEdit()
Result = QLineEdit()
add = QPushButton("Add")
sub = QPushButton("sub")
mul = QPushButton("mul")
div = QPushButton("div")

layout = QVBoxLayout()
layout2 = QHBoxLayout()

layout.addWidget(num1)
layout.addWidget(num2)



layout2.addWidget(add)
layout2.addWidget(sub)
layout2.addWidget(mul)
layout2.addWidget(div)


layout.addLayout(layout2)


num1.setAlignment(Qt.AlignRight)
num2.setAlignment(Qt.AlignRight)

num1.setPlaceholderText("Enter Your Number 1")
num2.setPlaceholderText("Enter Your Number 2")

#num1.setReadOnly(True)

window.setLayout(layout)


window.show()
app.exec()

#print(sys.version)
```
