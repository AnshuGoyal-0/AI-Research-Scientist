# Functions layer

```python
def add(a,b):
    return a+b

def sub(a,b):
    return a-b

def mul(a,b):
    return a*b

def div(a,b):
    if(a==0 or b == 0):
        print("division error")
    else:
        return a/b
```

# VARIABLES

```python
choice = ""
```

# Coding layer

# CLI CALCULATOR

```python

while True:
    choice = input(
        "\nEnter operation (add, sub, mul, div) or type stop:\n"
    ).lower()

    if "stop" in choice:
        print("Calculator stopped.")
        break

    try:
        num1 = int(input("Enter first number: "))
        num2 = int(input("Enter second number: "))
    except ValueError:
        print("Invalid number input. Try again.")
        continue

    # Keyword matching using ANY
    if any(word in choice for word in ["add", "sum", "plus", "+"]):
        print("Result:", add(num1, num2))

    elif any(word in choice for word in ["sub", "minus", "subtract", "-"]):
        print("Result:", sub(num1, num2))

    elif any(word in choice for word in ["mul", "into", "*"]):
        print("Result:", mul(num1, num2))

    elif any(word in choice for word in ["div", "division", "/", "bhag"]):
        print("Result:", div(num1, num2))

    else:
        print("Operation not recognized.")
```
