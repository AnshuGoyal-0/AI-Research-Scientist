## Import Layer

```python
import sys
from PySide6.QtWidgets import QApplication, QWidget, QLineEdit, QVBoxLayout, QPushButton, QHBoxLayout
from PySide6.QtCore import Qt
```

# Functions

```python
# ✅ FUNCTIONS (MUST BE BEFORE CONNECT)
def add():
    try:
        n1 = float(num1.text())
        n2 = float(num2.text())
        Result.setText(str(n1 + n2))
    except:
        Result.setText("Error")

def sub():
    try:
        n1 = float(num1.text())
        n2 = float(num2.text())
        Result.setText(str(n1 - n2))
    except:
        Result.setText("Error")

def mul():
    try:
        n1 = float(num1.text())
        n2 = float(num2.text())
        Result.setText(str(n1 * n2))
    except:
        Result.setText("Error")

def div():
    try:
        n1 = float(num1.text())
        n2 = float(num2.text())
        if n2 == 0:
            Result.setText("Divide by 0")
        else:
            Result.setText(str(n1 / n2))
    except:
        Result.setText("Error")
```

## Code Layer

```python
from PySide6.QtWidgets import *
from PySide6.QtCore import Qt

app = QApplication.instance()
if app is None:
    app = QApplication([])

window = QWidget()
window.setWindowTitle("Calculator Gui")

num1 = QLineEdit()
num2 = QLineEdit()
Result = QLineEdit()

add_btn = QPushButton("Add")
sub_btn = QPushButton("Sub")
mul_btn = QPushButton("Mul")
div_btn = QPushButton("Div")

# ✅ CONNECT BUTTONS
add_btn.clicked.connect(add)
sub_btn.clicked.connect(sub)
mul_btn.clicked.connect(mul)
div_btn.clicked.connect(div)

# Layout
layout = QVBoxLayout()
layout2 = QHBoxLayout()

layout.addWidget(num1)
layout.addWidget(num2)
layout.addWidget(Result)

layout2.addWidget(add_btn)
layout2.addWidget(sub_btn)
layout2.addWidget(mul_btn)
layout2.addWidget(div_btn)

layout.addLayout(layout2)

# UI settings
num1.setAlignment(Qt.AlignRight)
num2.setAlignment(Qt.AlignRight)
Result.setAlignment(Qt.AlignRight)

Result.setReadOnly(True)

window.setLayout(layout)
window.show()

app.exec()
```
